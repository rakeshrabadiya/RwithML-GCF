```r
#data Types
price<-90.50
age <- 20
#character(text)
name <- "Ehtesham"
#logical
passed <- TRUE
#Complex Number
number <- 3+2i

#check the data type using class()
print(class(price))
print(class(age))
print(class(name))
print(class(passed))
print(class(number))

```

```r
#Arithmatic Operation

a <- 10
b <- 3
print(a+b)
print(a-b)
print(a*b)
print(a/b)
print(a^b)
print(a%%b)
```

```r
#Celsius to Fahrenheit
celsius_to_fahrenheit <- function(celsius) {
  return((celsius * 9/5) + 32)
}

#Fahrenheit to Celsius
fahrenheit_to_celsius <- function(fahrenheit) {
  return((fahrenheit - 32) * 5/9)
}

```

```r
#comparison operators
a<-10
b<-20

print(a>b)
print(a<b)
print(a>=b)
print(a<=b)

#Logical Operator
age <-20
marks <- 80

#AND
#Both condition must be true
print(age>18 & marks>50)

# OR
#at least one condition must be true
print(age>18 | marks>90)

#NOT
#Changes TRUE to false and false to true
print(!(age>18))
```

```r
#if
age <- 20
if(age>=18){
  print("You are an adult")
}

#If-else
marks<-75
if(marks>=40){
  print("Pass")
}else{
  print("Fail")
}

#if else if else
marks<-85
if(marks>=90){
  print("Grade A+")
}else if(marks>=80){
  print("Grade A")
}else if(marks>=70){
  print("Grade B")
}else if(marks>=40){
  print("Grade c")
}else{
  print("Fail")
}
```

```r
#Write an R program to find the largest number among three numbers using if, else if, and else.

A<-20
B<-15
C<-25

if(A>B & A>C){
  print("A is Greater")
}else if(B>A & B>C){
  print("B is Greater")
}else{
  print("C is greater")
}

#Write an R program to check whether a number is positive, negative, or zero.
number<-15
if(number>0){
  print("Positive Number")
}else if(number<0){
  print("Negative Number")
}else{
  print("zero")
}
```

```r
#Access DataFrame
students<-data.frame(
  name=c("Ali","Ahmed","Sara","John"),
  age=c(20,21,19,22),
  marks=c(80,75,90,65)
)

print(students$name)
print(students$age)
print(students$marks)

#get first row
print(students[1,])
#get first two rows
print(students[1:2,])
#get first column
print(students[,1])

#Add Column
students<-data.frame(
  name=c("Ali","Ahmed","Sara","John"),
  marks=c(80,75,90,65)
)
print(students)


for(i in 1:nrow(students)){
  print(students$name[i])
  print(students$marks[i])
}
```

```r
students<-data.frame(
  name=c("ajay","bhavik","sumit"),
  age=c(20,21,22),
  rollno=c(1,2,3),
  attendance=c(80,74,1),
  result=c(80,60,20)
)
print(students)
barplot(
  students$result,
  names.arg = students$name,
  main = "Student Marks",
  xlab = "Students",
  ylab = "Marks"
)
```

```r
#Create a csv File
students<-data.frame(
  name=c("Ali","Ahmed","Sara","John"),
  age=c(20,21,19,22),
  marks=c(80,75,90,65)
)
#save the data as a csv file
write.csv(
  students,
  "students.csv",
  row.names = FALSE
)
print(students)
```
