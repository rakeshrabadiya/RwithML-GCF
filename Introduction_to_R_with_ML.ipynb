```r
# ====================================================
# EXPLORE AND CLEAN A SAMPLE DATASET IN R
# ====================================================

# ----------------------------------------------------
# STEP 1: CREATE A SAMPLE DATASET
# ----------------------------------------------------

# Create student dataset
students <- data.frame(
  name = c(" Ali ", "Ahmed", "Sara", "Ali", "John", "Neha"),
  age = c(20, 21, NA, 20, 22, 19),
  marks = c(80, 75, 90, 80, NA, 85),
  attendance = c(90, 85, 95, 90, 70, NA)
)

# Display the original dataset
#print("ORIGINAL DATASET:")
#print(students)


# ----------------------------------------------------
# STEP 2: EXPLORE THE DATASET
# ----------------------------------------------------

# Show first rows
print("FIRST ROWS:")
print(head(students))

# Show structure of the dataset
print("DATASET STRUCTURE:")
str(students)

# Show column names
print("COLUMN NAMES:")
print(names(students))

# Show number of rows
print("NUMBER OF ROWS:")
print(nrow(students))

# Show number of columns
print("NUMBER OF COLUMNS:")
print(ncol(students))

# Show statistical summary
print("DATA SUMMARY:")
print(summary(students))

# ----------------------------------------------------
# STEP 3: CHECK MISSING VALUES
# ----------------------------------------------------

# Check where missing values exist
print("MISSING VALUE CHECK:")
print(is.na(students))

# Count missing values in each column
print("MISSING VALUES BY COLUMN:")
print(colSums(is.na(students)))

# Count total missing values
print("TOTAL MISSING VALUES:")
print(sum(is.na(students)))


# ----------------------------------------------------
# STEP 4: HANDLE MISSING VALUES
# ----------------------------------------------------

# Replace missing age with the average age
students$age[is.na(students$age)] <- mean(
  students$age,
  na.rm = TRUE
)

# Replace missing marks with the average marks
students$marks[is.na(students$marks)] <- mean(
  students$marks,
  na.rm = TRUE
)

# Replace missing attendance with the average attendance
students$attendance[is.na(students$attendance)] <- mean(
  students$attendance,
  na.rm = TRUE
)


# ----------------------------------------------------
# STEP 5: REMOVE EXTRA SPACES
# ----------------------------------------------------

# Remove spaces from the beginning and end of names
students$name <- trimws(students$name)

# ----------------------------------------------------
# STEP 6: FIND DUPLICATES
# ----------------------------------------------------

# Check duplicate rows
print("DUPLICATE ROW CHECK:")
print(duplicated(students))

# Count duplicate rows
print("NUMBER OF DUPLICATES:")
print(sum(duplicated(students)))


# ----------------------------------------------------
# STEP 7: REMOVE DUPLICATES
# ----------------------------------------------------

# Remove duplicate rows
students <- students[!duplicated(students), ]

# ----------------------------------------------------
# STEP 8: CHECK CLEAN DATA
# ----------------------------------------------------

print("CLEAN DATASET:")
print(students)

# Check missing values again
print("MISSING VALUES AFTER CLEANING:")
print(colSums(is.na(students)))

# Check duplicates again
print("DUBLITCATES AFTER CLEANING:")
print(sum(duplicated(students)))

# Show final structure
print("FINAL DATASET STRUCTURE:")
str(students)

# ----------------------------------------------------
# STEP 9: SAVE CLEAN DATASET
# ----------------------------------------------------

# Save cleaned dataset as a CSV file
write.csv(
  students,
  "clean_students.csv",
  row.names = FALSE
)

print("Clean dataset saved as: clean_students.csv")
```

```r
# Create a small student dataset
students <- data.frame(
  Name = c("Ali", "Sara", "Rahul", "Aman", "Neha",
           "Zoya", "Ravi", "Priya", "Arjun", "Simran"),
  Marks = c(45, 52, 58, 62, 65, 70, 72, 78, 85, 92)
)

# View the dataset
print(students)

# Create histogram of student marks
hist(
  students$Marks,                         # Select Marks column
  breaks = 5,                             # Suggestion for number of bins
  main = "Distribution of Student Marks", # Graph title
  xlab = "Marks",                         # X-axis label
  ylab = "Number of Students"             # Y-axis label
)
```

```r
# Add study hours to our students dataset.
# Each value represents how many hours a student studied.

students$StudyHours <- c(
  5,
  3,
  7,
  2,
  6,
  4,
  8,
  3,
  5,
  7
)


# Create scatter plot.
# A scatter plot shows the relationship between two variables.

plot(
  students$StudyHours, # X-axis: study hours
  students$Marks,      # Y-axis: student marks

  # Graph title
  main = "Study Hours vs Marks",

  # X-axis label
  xlab = "Study Hours",

  # Y-axis label
  ylab = "Marks",

)

# Each point represents one student.
#
# Example:
# StudyHours = 5 and Marks = 45
# means one student studied 5 hours and got 45 marks.
#
# Scatter plot helps us see whether study hours
# and marks have a relationship.
```

```r
# ===================================================================
# PART 28: PIE CHART
# ===================================================================

# Create student dataset

students <- data.frame(
  Name = c(
    "Ali", "Sara", "Rahul", "Aman", "Neha",
    "Zoya", "Ravi", "Priya", "Arjun", "Simran"
  ),

  Marks = c(
    45, 52, 58, 62, 65,
    70, 72, 78, 85, 92
  ),

  City = c(
    "Delhi", "Mumbai", "Delhi", "Pune", "Mumbai",
    "Delhi", "Pune", "Mumbai", "Delhi", "Pune"
  )
)


# Count students from each city
# table() counts how many students belong to each city

city_count <- table(students$City)


# Display the city counts

city_count

# Create pie chart

pie(
  city_count,
  labels = names(city_count),         # Display city names on the pie chart
  main = "Students by City"           # Add chart title
)
```
