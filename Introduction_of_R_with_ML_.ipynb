```r
# =========================================
# COMPLETE DATA CLEANING
# =========================================

# Create dataset
students <- data.frame(
  name = c(
    " Ali ",
    "Ahmed",
    "Sara",
    " Ali ",
    "John"
  ),

  age = c(
    20,
    21,
    NA,
    20,
    22
  ),

  marks = c(
    80,
    75,
    90,
    80,
    NA
  )
)

# -----------------------------------------
# STEP 1: VIEW ORIGINAL DATA
# -----------------------------------------

print(students)

# -----------------------------------------
# STEP 2: REMOVE EXTRA SPACES
# -----------------------------------------

students$name <- trimws(students$name)

# -----------------------------------------
# STEP 3: CHECK MISSING VALUES
# -----------------------------------------

print(colSums(is.na(students)))

# -----------------------------------------
# STEP 4: REPLACE MISSING AGE
# -----------------------------------------

students$age[
  is.na(students$age)
] <- mean(
  students$age,
  na.rm = TRUE
)

# -----------------------------------------
# STEP 5: REPLACE MISSING MARKS
# -----------------------------------------

students$marks[
  is.na(students$marks)
] <- mean(
  students$marks,
  na.rm = TRUE
)

# -----------------------------------------
# STEP 6: REMOVE DUPLICATES
# -----------------------------------------

students <- unique(students)

# -----------------------------------------
# STEP 7: VIEW CLEAN DATA
# -----------------------------------------
print("unique")
print(students)

# -----------------------------------------
# STEP 8: CHECK AGAIN
# -----------------------------------------

print(colSums(is.na(students)))

# Check structure
str(students)
```

```r
# ============================================================
# EXPLORE AND CLEAN A SAMPLE DATASET IN R
# ============================================================

# ------------------------------------------------------------
# STEP 1: CREATE A SAMPLE DATASET
# ------------------------------------------------------------

# Create student dataset
students <- data.frame(
  name = c(" Ali ", "Ahmed", "Sara", "Ali", "John", "Neha"),
  age = c(20, 21, NA, 20, 22, 19),
  marks = c(80, 75, 90, 80, NA, 85),
  attendance = c(90, 85, 95, 90, 70, NA)
)

# Display the original dataset
print("ORIGINAL DATASET:")
print(students)


# ------------------------------------------------------------
# STEP 2: EXPLORE THE DATASET
# ------------------------------------------------------------

# Show first rows
print("FIRST ROWS:")
print(head(students))

# Show structure of the dataset
print("DATASET STRUCTURE:")
str(students)

# Show column names
print("COLUMN NAMES:")
print(names(students))

# Show number of rows
print("NUMBER OF ROWS:")
print(nrow(students))

# Show number of columns
print("NUMBER OF COLUMNS:")
print(ncol(students))

# Show statistical summary
print("DATA SUMMARY:")
print(summary(students))


# ------------------------------------------------------------
# STEP 3: CHECK MISSING VALUES
# ------------------------------------------------------------

# Check where missing values exist
print("MISSING VALUE CHECK:")
print(is.na(students))

# Count missing values in each column
print("MISSING VALUES BY COLUMN:")
print(colSums(is.na(students)))

# Count total missing values
print("TOTAL MISSING VALUES:")
print(sum(is.na(students)))


# ------------------------------------------------------------
# STEP 4: HANDLE MISSING VALUES
# ------------------------------------------------------------

# Replace missing age with the average age
students$age[is.na(students$age)] <- mean(
  students$age,
  na.rm = TRUE
)

# Replace missing marks with the average marks
students$marks[is.na(students$marks)] <- mean(
  students$marks,
  na.rm = TRUE
)

# Replace missing attendance with the average attendance
students$attendance[is.na(students$attendance)] <- mean(
  students$attendance,
  na.rm = TRUE
)


# ------------------------------------------------------------
# STEP 5: REMOVE EXTRA SPACES
# ------------------------------------------------------------

# Remove spaces from the beginning and end of names
students$name <- trimws(students$name)


# ------------------------------------------------------------
# STEP 6: FIND DUPLICATES
# ------------------------------------------------------------

# Check duplicate rows
print("DUPLICATE ROW CHECK:")
print(duplicated(students))

# Count duplicate rows
print("NUMBER OF DUPLICATES:")
print(sum(duplicated(students)))


# ------------------------------------------------------------
# STEP 7: REMOVE DUPLICATES
# ------------------------------------------------------------

# Remove duplicate rows
students <- students[!duplicated(students), ]


# ------------------------------------------------------------
# STEP 8: CHECK CLEAN DATA
# ------------------------------------------------------------

print("CLEAN DATASET:")
print(students)

# Check missing values again
print("MISSING VALUES AFTER CLEANING:")
print(colSums(is.na(students)))

# Check duplicates again
print("DUPLICATES AFTER CLEANING:")
print(sum(duplicated(students)))

# Show final structure
print("FINAL DATASET STRUCTURE:")
str(students)


# ------------------------------------------------------------
# STEP 9: SAVE CLEAN DATASET
# ------------------------------------------------------------

# Save cleaned dataset as a CSV file
write.csv(
  students,
  "clean_students.csv",
  row.names = FALSE
)

print("Clean dataset saved as: clean_students.csv")
```

```r
# ============================================================
# DESCRIPTIVE STATISTICS IN R
# Mean, Median, Mode, Variance and Standard Deviation
# ============================================================

# Create sample data
marks <- c(10, 20, 20, 30, 40)

# -----------------------------------------
# MEAN
# -----------------------------------------

mean_value <- mean(marks)

print("Mean:")
print(mean_value)


# -----------------------------------------
# MEDIAN
# -----------------------------------------

median_value <- median(marks)

print("Median:")
print(median_value)


# -----------------------------------------
# MODE
# -----------------------------------------

# Count frequency of each value
frequency <- table(marks)

# Find the value occurring most often
mode_value <- names(frequency)[which.max(frequency)]

print("Mode:")
print(mode_value)


# -----------------------------------------
# VARIANCE
# -----------------------------------------

variance_value <- var(marks)

print("Variance:")
print(variance_value)


# -----------------------------------------
# STANDARD DEVIATION
# -----------------------------------------

sd_value <- sd(marks)

print("Standard Deviation:")
print(sd_value)


# -----------------------------------------
# COMPLETE SUMMARY
# -----------------------------------------

print("Complete Summary:")
print(summary(marks))
```

```r
# ============================================================
# INTRODUCTION TO CARET
# ============================================================

# caret is an R package used for Machine Learning.
# It helps us train, test and evaluate ML models.

# ------------------------------------------------------------
# STEP 1: Install caret
# ------------------------------------------------------------

# install.packages() is used to install an R package.
# Run this only when the package is not already installed.

install.packages("caret")


# ------------------------------------------------------------
# STEP 2: Load caret
# ------------------------------------------------------------

# library() loads the package so that we can use its functions.

library(caret)


# ------------------------------------------------------------
# STEP 3: Create a small dataset
# ------------------------------------------------------------

# We have two columns:
# StudyHours = number of hours a student studies
# Marks      = marks obtained by the student

students <- data.frame(
  StudyHours = c(1, 2, 3, 4, 5, 6, 7, 8),
  Marks = c(35, 40, 45, 50, 55, 65, 70, 80)
)

# Display our dataset
students


# ------------------------------------------------------------
# STEP 4: Split data into Training and Testing data
# ------------------------------------------------------------

# Machine Learning normally uses:
# Training data -> used to teach the model
# Testing data  -> used to check the model

# set.seed() makes the random split reproducible.
set.seed(123)

# createDataPartition() is a caret function.
# p = 0.8 means approximately 80% data for training.

train_index <- createDataPartition(
  students$Marks,
  p = 0.8,
  list = FALSE
)

# Create training data
train_data <- students[train_index, ]

# Create testing data
test_data <- students[-train_index, ]


# Show training data
train_data

# Show testing data
test_data


# ------------------------------------------------------------
# STEP 5: Train a Linear Regression model
# ------------------------------------------------------------

# train() is one of the important functions of caret.
#
# Marks ~ StudyHours means:
# We want to predict Marks
# using StudyHours.

model <- train(
  Marks ~ StudyHours,
  data = train_data,
  method = "lm"
)

# "lm" means Linear Regression.


# ------------------------------------------------------------
# STEP 6: See the trained model
# ------------------------------------------------------------

# Print the model information
model


# ------------------------------------------------------------
# STEP 7: Make predictions
# ------------------------------------------------------------

# predict() uses our trained model
# to predict marks for testing data.

predictions <- predict(
  model,
  newdata = test_data
)

# Display predictions
predictions


# ------------------------------------------------------------
# STEP 8: Compare actual and predicted values
# ------------------------------------------------------------

# Actual marks
test_data$Marks

# Predicted marks
predictions


# ------------------------------------------------------------
# STEP 9: Calculate model performance
# ------------------------------------------------------------

# postResample() is a caret function.
# It gives common regression performance measures.

postResample(
  pred = predictions,
  obs = test_data$Marks
)

# The result includes:
# RMSE = Root Mean Squared Error
# Rsquared = R-squared
# MAE = Mean Absolute Error


# ============================================================
# END
# ============================================================
```

```r
# ============================================================
# LINEAR REGRESSION USING mlr3
# PROJECT: Predict Student Exam Result from Study Hours
# ============================================================


# ------------------------------------------------------------
# STEP 1: INSTALL REQUIRED PACKAGES
# ------------------------------------------------------------

# Install mlr3 package
# mlr3 is a machine learning framework used to create
# tasks, learners, predictions and evaluate models.
install.packages("mlr3")

# Install mlr3learners package
# This package provides different machine learning algorithms,
# including Linear Regression.
install.packages("mlr3learners")


# ------------------------------------------------------------
# STEP 2: LOAD REQUIRED PACKAGES
# ------------------------------------------------------------

# Load mlr3 package into R
library(mlr3)

# Load mlr3learners package into R
library(mlr3learners)


# ------------------------------------------------------------
# STEP 3: CREATE STUDENT DATA
# ------------------------------------------------------------

# data.frame() is used to create a table of data.
#
# StudyHours = number of hours a student studies
#
# Result = marks obtained by the student
#
# Here, StudyHours is our INPUT/FEATURE.
# Result is our OUTPUT/TARGET.

data <- data.frame(

  # Study hours of 10 students
  StudyHours = c(
    1, 2, 2, 3, 4,
    5, 5, 6, 7, 8
  ),

  # Exam marks of the same 10 students
  Result = c(
    35, 40, 45, 50, 55,
    65, 68, 72, 80, 88
  )
)


# ------------------------------------------------------------
# STEP 4: DISPLAY THE DATA
# ------------------------------------------------------------

# print() displays the data in the R console.
print(data)


# ------------------------------------------------------------
# STEP 5: UNDERSTAND OUR VARIABLES
# ------------------------------------------------------------

# StudyHours = independent variable / predictor
#
# Result = dependent variable / target
#
# Our goal is:
#
# Study Hours  --->  Linear Regression  --->  Exam Result
#
# Example:
#
# 6 hours study ---> Model ---> predicted marks


# ------------------------------------------------------------
# STEP 6: CREATE A REGRESSION TASK
# ------------------------------------------------------------

# TaskRegr$new() creates a regression task in mlr3.
#
# A regression task contains:
# 1. Input features
# 2. Target/output variable
# 3. Dataset
#
# id = unique name of our task
#
# backend = our dataset
#
# target = column that we want to predict

task <- TaskRegr$new(
  id = "student_result",
  backend = data,
  target = "Result"
)


# ------------------------------------------------------------
# STEP 7: DISPLAY THE TASK
# ------------------------------------------------------------

# print(task) shows information about our regression task.
#
# It tells us:
# - Number of observations
# - Number of features
# - Target variable
# - Feature names

print(task)


# ------------------------------------------------------------
# STEP 8: CREATE LINEAR REGRESSION LEARNER
# ------------------------------------------------------------

# lrn() is used to create a machine learning learner.
#
# "regr.lm" means:
#
# regr = regression
# lm   = Linear Model
#
# Therefore:
# regr.lm = Linear Regression algorithm

learner <- lrn("regr.lm")


# ------------------------------------------------------------
# STEP 9: DISPLAY THE LEARNER
# ------------------------------------------------------------

# print() displays information about our
# Linear Regression learner.

print(learner)


# ------------------------------------------------------------
# STEP 10: TRAIN THE MODEL
# ------------------------------------------------------------

# $train() trains the machine learning model.
#
# The model looks at:
#
# StudyHours ---> Result
#
# and learns the relationship between them.

learner$train(task)


# ------------------------------------------------------------
# STEP 11: DISPLAY THE TRAINED MODEL
# ------------------------------------------------------------

# learner$model contains the actual Linear Regression model.
#
# It shows information such as:
# - Intercept
# - StudyHours coefficient
# - Residual information
# - Statistical information

print(learner$model)


# ------------------------------------------------------------
# STEP 12: MAKE PREDICTIONS
# ------------------------------------------------------------

# $predict() uses the trained model to make predictions.
#
# Here we are predicting results for the students
# already present in our dataset.

prediction <- learner$predict(task)


# ------------------------------------------------------------
# STEP 13: DISPLAY PREDICTED VALUES
# ------------------------------------------------------------

# prediction$response contains the predicted exam marks.

print(prediction$response)


# ------------------------------------------------------------
# STEP 14: COMPARE ACTUAL AND PREDICTED VALUES
# ------------------------------------------------------------

# Create a new table containing:
#
# StudyHours       = student's study hours
# ActualResult     = actual exam marks
# PredictedResult  = marks predicted by the model

result <- data.frame(

  # Original study hours
  StudyHours = data$StudyHours,

  # Actual marks
  ActualResult = data$Result,

  # Predicted marks
  # round(..., 2) keeps only 2 decimal places
  PredictedResult = round(
    prediction$response,
    2
  )
)


# Display comparison table
print(result)


# ------------------------------------------------------------
# STEP 15: CALCULATE RMSE
# ------------------------------------------------------------

# RMSE means:
# Root Mean Squared Error
#
# It tells us how far our predictions are from
# the actual values.
#
# Smaller RMSE = smaller prediction error.

rmse <- prediction$score(
  msr("regr.rmse")
)

# Display RMSE
print(rmse)


# ------------------------------------------------------------
# STEP 16: CALCULATE R-SQUARED
# ------------------------------------------------------------

# R-squared (R²) tells us how well the model
# explains the variation in the target variable.
#
# R² is generally between 0 and 1.
#
# A value closer to 1 means the model explains
# more of the variation in the data.

rsq <- prediction$score(
  msr("regr.rsq")
)

# Display R-squared
print(rsq)


# ------------------------------------------------------------
# STEP 17: PREDICT RESULT FOR A NEW STUDENT
# ------------------------------------------------------------

# Suppose a new student studies for 6 hours.
#
# We want to predict the student's exam result.

new_student <- data.frame(
  StudyHours = 6
)


# ------------------------------------------------------------
# STEP 18: MAKE PREDICTION FOR NEW DATA
# ------------------------------------------------------------

# predict_newdata() uses our trained model
# to predict the result for new data.

new_prediction <- learner$predict_newdata(
  newdata = new_student
)


# ------------------------------------------------------------
# STEP 19: DISPLAY NEW STUDENT PREDICTION
# ------------------------------------------------------------

# response contains the predicted value.

cat(
  "Predicted Exam Result:",
  round(new_prediction$response, 2)
)


# ============================================================
# END OF LINEAR REGRESSION PROJECT
# ============================================================
```

```r
# ============================================================
# LINEAR REGRESSION IN R
# ============================================================

# Linear Regression:
# Used when the output is a continuous numerical value.
#
# Example:
# Study Hours -> Marks
#
# Here:
# X = StudyHours
# Y = Marks


# ------------------------------------------------------------
# STEP 1: Create sample dataset
# ------------------------------------------------------------

students <- data.frame(
  StudyHours = c(1, 2, 3, 4, 5, 6, 7, 8),
  Marks = c(35, 40, 50, 55, 65, 70, 80, 85)
)

# Display the dataset
students


# ------------------------------------------------------------
# STEP 2: Visualize the data
# ------------------------------------------------------------

plot(
  students$StudyHours,
  students$Marks,
  main = "Study Hours vs Marks",
  xlab = "Study Hours",
  ylab = "Marks",
  pch = 19
)

# We can see that marks generally increase
# when study hours increase.


# ------------------------------------------------------------
# STEP 3: Create Linear Regression Model
# ------------------------------------------------------------

model <- lm(
  Marks ~ StudyHours,
  data = students
)

# lm() means Linear Model.
#
# Marks ~ StudyHours means:
# Predict Marks using StudyHours.


# ------------------------------------------------------------
# STEP 4: See model information
# ------------------------------------------------------------

summary(model)

# Important things in the output:
#
# Coefficients:
#   (Intercept) = b0
#   StudyHours  = b1
#
# R-squared:
#   Tells us how well the model explains the data.


# ------------------------------------------------------------
# STEP 5: Draw regression line
# ------------------------------------------------------------

plot(
  students$StudyHours,
  students$Marks,
  main = "Linear Regression",
  xlab = "Study Hours",
  ylab = "Marks",
  pch = 19
)

# Add the regression line
abline(model)


# ------------------------------------------------------------
# STEP 6: Make prediction
# ------------------------------------------------------------

# Suppose a student studies for 9 hours.

new_student <- data.frame(
  StudyHours = 9
)

prediction <- predict(
  model,
  newdata = new_student
)

prediction

# The model predicts the expected marks
# for a student who studies 9 hours.
```

```r
# ============================================
# DECISION TREE - SIMPLE EXAMPLE IN R
# ============================================

# ============================================
# STEP 1: CREATE DATASET
# ============================================

# We have:
# StudyHours = Number of hours a student studies
# Attendance = Attendance percentage
# Result = Pass or Fail

students <- data.frame(
  StudyHours = c(2, 3, 4, 5, 6, 7, 8, 9, 10),
  Attendance = c(60, 65, 70, 75, 80, 82, 85, 90, 95),
  Result = c(
    "Fail", "Fail", "Fail",
    "Pass", "Pass", "Pass",
    "Pass", "Pass", "Pass"
  )
)

# ============================================
# STEP 2: CONVERT RESULT INTO FACTOR
# ============================================

# Decision Tree classification works with categories.
# "Pass" and "Fail" are categories.

students$Result <- as.factor(students$Result)



# ============================================
# STEP 3: CREATE DECISION TREE MODEL
# ============================================

# Result is the target/output variable.
# StudyHours and Attendance are input variables.

model <- rpart(
  Result ~ StudyHours + Attendance,
  data = students,
  method = "class"
)


# ============================================
# STEP 5: PREDICT NEW STUDENT
# ============================================

# Suppose a new student has:
# Study Hours = 6
# Attendance = 80%

new_student <- data.frame(
  StudyHours = 1,
  Attendance = 8
)

# Predict whether the student will Pass or Fail
prediction <- predict(
  model,
  new_student,
  type = "class"
)

# Display prediction
print(prediction)
```

```r
# ==========================================
# K-Nearest Neighbors (KNN) in R
# ==========================================

# Install the class package if it is not installed
# Run this only once
install.packages("class")

# Load the class package
library(class)


# ------------------------------------------
# STEP 1: Create simple student data
# ------------------------------------------

# Study hours of students
StudyHours <- c(1, 2, 3, 4, 5, 6, 7, 8)

# Marks obtained by students
Marks <- c(30, 35, 40, 50, 55, 65, 75, 85)

# Result of each student
Result <- c(
  "Fail", "Fail", "Fail", "Pass",
  "Pass", "Pass", "Pass", "Pass"
)


# ------------------------------------------
# STEP 2: Create a data frame
# ------------------------------------------

students <- data.frame(
  StudyHours,
  Marks,
  Result
)

# ------------------------------------------
# STEP 3: Separate input and output
# ------------------------------------------

# Input features
# KNN uses StudyHours and Marks to make prediction
X <- students[, c("StudyHours", "Marks")]

# Output/target
# This is what we want to predict
Y <- students$Result


# ------------------------------------------
# STEP 4: Create a new student
# ------------------------------------------

# New student studied for 5 hours
# and obtained 60 marks
new_student <- data.frame(
  StudyHours = 5,
  Marks = 60
)


# ------------------------------------------
# STEP 5: Apply KNN
# ------------------------------------------

# k = 3 means:
# Look at the 3 nearest students
# and use majority voting
prediction <- knn(
  train = X,              # Training input data
  test = new_student,     # New data to predict
  cl = Y,                 # Actual classes/results
  k = 3                   # Number of nearest neighbors
)


# ------------------------------------------
# STEP 6: Display prediction
# ------------------------------------------

print(prediction)


# ------------------------------------------
# FINAL RESULT
# ------------------------------------------

# The prediction will be:
# "Pass"
#
# Because the 3 nearest students
# mostly belong to the Pass class.
```

```r
# ============================================
# K-MEANS CLUSTERING IN R
# ============================================

# Step 1: Create sample student data
# StudyHours = number of hours students study
# Marks = marks obtained by students

students <- data.frame(
  StudyHours = c(1, 2, 2, 3, 4, 5, 6, 7, 8, 9),
  Marks = c(35, 40, 42, 48, 55, 60, 68, 75, 82, 90)
)

# Display the data
print(students)


# Step 2: Select the columns that we want to use
# K-Means works with numerical data

data_for_clustering <- students[, c("StudyHours", "Marks")]


# Step 3: Apply K-Means
# centers = 3 means we want 3 clusters

set.seed(123)  # Makes the result reproducible

kmeans_model <- kmeans(data_for_clustering, centers = 3)


# Step 4: Show which cluster each student belongs to

print(kmeans_model$cluster)


# Step 5: Add the cluster number to our original data

students$Cluster <- kmeans_model$cluster


# Step 6: Display final result

print(students)


# Step 7: Display the center of each cluster
# These are called centroids

print(kmeans_model$centers)


# Step 8: Visualize the clusters

plot(
  students$StudyHours,
  students$Marks,
  col = students$Cluster,       # Different color for each cluster
  pch = 19,                     # Solid circle
  xlab = "Study Hours",
  ylab = "Marks",
  main = "K-Means Clustering"
)

# Add cluster centers to the graph
points(
  kmeans_model$centers[, "StudyHours"],
  kmeans_model$centers[, "Marks"],
  pch = 8,
  cex = 2
)
```

```r
# ============================================
# ANOMALY DETECTION IN R
# Using Z-Score
# ============================================

# Step 1: Create some sample data
# Most values are normal, but 100 is an unusual value

data <- c(10, 12, 11, 13, 12, 14, 11, 15, 13, 12, 100)

# Display the data
print(data)


# Step 2: Calculate the mean
# Mean = average value

mean_value <- mean(data)

# Display mean
print(mean_value)


# Step 3: Calculate standard deviation

sd_value <- sd(data)

# Display standard deviation
print(sd_value)


# Step 4: Calculate Z-score for every value
# Z-score tells us how far a value is from the mean

z_score <- (data - mean_value) / sd_value

# Display Z-scores
print(z_score)


# Step 5: Find anomalies
# If the absolute Z-score is greater than 2,
# we consider it an anomaly

anomaly <- abs(z_score) > 2


# Step 6: Display the anomaly result

print(anomaly)


# Step 7: Display only the anomaly values

print(data[anomaly])


# Step 8: Display normal values

print(data[!anomaly])


# ============================================
# FINAL RESULT
# ============================================

cat("Anomaly values are:\n")
print(data[anomaly])
```

```r
# ============================================
# HIERARCHICAL CLUSTERING IN R
# ============================================

# Step 1: Create sample student data
# StudyHours = hours studied
# Marks = marks obtained

students <- data.frame(
  StudyHours = c(1, 2, 2, 3, 7, 8, 9, 10),
  Marks = c(35, 40, 42, 45, 70, 75, 85, 90)
)

# Display the data
print(students)


# Step 2: Calculate distance between students
# dist() calculates the distance between data points

distance <- dist(students)

# Display the distances
print(distance)


# Step 3: Perform hierarchical clustering
# method = "complete" means complete linkage
# It decides the distance between clusters
# based on the farthest pair of points

hc <- hclust(distance, method = "complete")


# Step 4: Display the clustering result

print(hc)


# Step 5: Draw the dendrogram
# A dendrogram is a tree showing how
# the data points are grouped

plot(
  hc,
  main = "Hierarchical Clustering Dendrogram",
  xlab = "Students",
  ylab = "Distance"
)


# Step 6: Create 2 clusters
# k = 2 means we want to divide the data
# into 2 final groups

clusters <- cutree(hc, k = 2)


# Step 7: Add cluster numbers to the data

students$Cluster <- clusters


# Step 8: Display the final result

print(students)
```

```r
# Install package (run only once)
install.packages("isotree")

# Load library
library(isotree)

# -----------------------------
# 1. Create sample data
# -----------------------------

data <- data.frame(
  Age = c(20, 21, 22, 23, 24, 25, 26, 27, 28, 80),
  Salary = c(25000, 27000, 26000, 28000, 30000,
             32000, 31000, 33000, 35000, 200000)
)

# Display data
print(data)


# -----------------------------
# 2. Create Isolation Forest model
# -----------------------------

model <- isolation.forest(
  data,
  ntrees = 100
)


# -----------------------------
# 3. Calculate anomaly score
# -----------------------------

scores <- predict(model, data)


# -----------------------------
# 4. Add score to data
# -----------------------------

data$Anomaly_Score <- scores


# -----------------------------
# 5. Identify anomalies
# -----------------------------

# Higher score = more unusual
threshold <- 0.65

data$Anomaly <- ifelse(
  data$Anomaly_Score > threshold,
  "Anomaly",
  "Normal"
)


# -----------------------------
# 6. Display final result
# -----------------------------

print(data)


# -----------------------------
# 7. Display only anomalies
# -----------------------------

print(data[data$Anomaly == "Anomaly", ])
```
